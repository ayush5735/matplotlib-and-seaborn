```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 
```

matplotlib basics

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("hello")
plt.xlabel("x axis")
plt.ylabel("y axis")
```

subplots

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

object oriented way

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,1,1]) 
#  last ke do factor width aur hieght hai
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.1,0.3,0.3]) 
axis2.plot(x,y)
# surwatr ki jo so hai wo left aur right ke liye hai
```

types of plot

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

saving plots

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basicplot.png")
```
